# move and remov logic

```python
import os
import re
import shutil

root_path   = r"E:\year4\spectralis\final"   # folder to scan
output_path = r"E:\mactel\spectralis"        # destination

pattern = re.compile(r"^[369]\d{3}$")     # exactly 4 digits, starting with 3, 6 or 9

DRY_RUN = False      # True = preview only, nothing is touched. False = execute.
MODE    = "move"    # "move" = originals are deleted after transfer.

os.makedirs(output_path, exist_ok=True)

found = 0
done = 0
skipped = 0

for root, dirs, files in os.walk(root_path):
    matched = [d for d in dirs if pattern.match(d)]

    for dname in matched:
        src = os.path.join(root, dname)
        rel = os.path.relpath(src, root_path)
        dst = os.path.join(output_path, rel)

        found += 1
        print(f"{dname}", flush=True)
        print(f"   from: {src}", flush=True)
        print(f"   to  : {dst}", flush=True)

        if os.path.exists(dst):
            skipped += 1
            print("   [SKIP] destination exists — source NOT removed", flush=True)
            continue

        if DRY_RUN:
            print(f"   [DRY] would {MODE} (source would be deleted)", flush=True)
        else:
            os.makedirs(os.path.dirname(dst), exist_ok=True)
            if MODE == "move":
                shutil.move(src, dst)
                print("   -> moved (source deleted)", flush=True)
            else:
                shutil.copytree(src, dst)
                print("   -> copied", flush=True)
            done += 1

    for d in matched:
        dirs.remove(d)

print(f"\nfound: {found}, {MODE}d: {done}, skipped: {skipped}. DRY_RUN = {DRY_RUN}", flush=True)
```

```python
import os
import re
import shutil

root_path   = r"E:\year4\optomed\final"          # folder to scan
output_path = r"E:\mactel\optomed"     # destination


pattern = re.compile(r"^[369]\d{3}$")   # exactly 4 digits, starting with 3, 6 or 9

DRY_RUN = False     # True = preview only, nothing is touched. False = execute.
MODE    = "copy"   # "copy" = keep originals.  "move" = remove originals.

pattern = re.compile(r"^[369]\d{3}$")   # exactly 4 digits, starting with 3, 6 or 9

os.makedirs(output_path, exist_ok=True)

found = 0
done = 0
for root, dirs, files in os.walk(root_path):
    matched = [d for d in dirs if pattern.match(d)]

    for dname in matched:
        src = os.path.join(root, dname)

        # keep the parent path so identical IDs from different
        # locations don't collide at the destination
        rel = os.path.relpath(src, root_path)
        dst = os.path.join(output_path, rel)

        found += 1
        print(f"{dname}", flush=True)
        print(f"   from: {src}", flush=True)
        print(f"   to  : {dst}", flush=True)

        if os.path.exists(dst):
            print("   [SKIP] already exists at destination", flush=True)
            continue

        if not DRY_RUN:
            os.makedirs(os.path.dirname(dst), exist_ok=True)
            if MODE == "move":
                shutil.move(src, dst)
                print("   -> moved", flush=True)
            else:
                shutil.copytree(src, dst)
                print("   -> copied", flush=True)
            done += 1

    # don't descend into folders already handled (whole tree is copied anyway)
    for d in matched:
        dirs.remove(d)

print(f"\nfound: {found}, {MODE}d: {done}. DRY_RUN = {DRY_RUN}", flush=True)
```
