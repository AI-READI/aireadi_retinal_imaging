# List Maestro2 intersecting iDs

```python
import pandas as pd

df = pd.read_csv(r"C:\path...\dicom_ids_for_m2.csv", dtype=str, keep_default_na=False)
ids= [ 1234, 1234 ]  # list of IDs that con
pd.set_option("display.max_colwidth", None)
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)


df["num"] = df["patient_id"].str.extract(r"(\d+)\s*$")[0].str.lstrip("0")

results = []
for raw in ids:
    key = (str(int(raw)) if isinstance(raw, float) else str(raw).strip()).lstrip("0")
    m = df[df["num"] == key]
    results.append({
        "search_id": raw,
        "found": bool(len(m)),
        "matched_ids": ", ".join(sorted(m["patient_id"].unique())),
        "batches": ", ".join(sorted(m["batch_folder"].unique())),
        "sub_batches": ", ".join(sorted(m["sub_folder"].unique())),
    })

results = pd.DataFrame(results)
print(f"Found: {results.found.sum()} / {len(results)}")
results.to_csv("id_check_for_maestro2.csv", index=False)
results
```
