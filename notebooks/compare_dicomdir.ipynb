```python
from pydicom import dcmread
from pydicom.fileset import FileSet
```

```python
original_path = r"D:\year3+raw\spectralis-s\UCSD_Spectralis_20260501-20260531\DICOMDIR"
recreated_path = r"D:\year3+raw\spectralis-s\UCSD_Spectralis_20260501-20260531\DICOMDIR_generated"
```

```python
original_ds = dcmread(original_path)
original_fs = FileSet(original_ds)
recreated_ds = dcmread(recreated_path)
recreated_fs = FileSet(recreated_ds)
```

```python
from pathlib import Path

Path("original_fs.txt").write_text(str(original_fs), encoding="utf-8")
Path("recreated_fs.txt").write_text(str(recreated_fs), encoding="utf-8")
```
